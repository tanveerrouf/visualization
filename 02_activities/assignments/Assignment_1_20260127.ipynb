```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests

np.random.seed(613)
x1 = np.arange(50)
y1 = np.random.randint(0, 75,50)
x2 = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y2 = np.array([110, 180, 240, 99, 220])
```

```python
fig,(ax1, ax2) = plt.subplots(ncols=2,
 nrows=1,
 figsize=(10, 3))

ax1.scatter(x1,y1)
ax2.bar(x2,y2)
```

```python
# Annotations can be added like this:
fig,(ax1, ax2) = plt.subplots(ncols=2,
 nrows=1,
 figsize=(10, 3))

ax1.scatter(x1,y1)
ax1.annotate('This is important', xy=(x1[24], y1[24]), xytext=(x1[24]+5, y1[24]+10),
            arrowprops=dict(facecolor='black', shrink=0.05))
ax2.bar(x2,y2)
```

```python
fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'],
 ['ax2', 'ax3']],
 figsize=(20, 10))

someaxes["ax1"].scatter(x1,y1)
label = 'Important Point'
someaxes["ax1"].annotate(label, xy=(x1[24], y1[24]), xytext=(x1[24]+5, y1[24]+10),  
            arrowprops=dict(facecolor='green', shrink=0.05))

someaxes["ax2"].bar(x2,y2)
label = 'Highest Value'
someaxes["ax2"].annotate(label, xy=(x2[2], y2[2]), xytext=(2, y2[2]+30),  
            arrowprops=dict(facecolor='violet', shrink=0.05))

someaxes["ax3"].plot(x1,y1)
label = 'Mid Point'
someaxes["ax3"].annotate(label, xy=(x1[25], y1[25]), xytext=(x1[25]+5, y1[25]+10),  
            arrowprops=dict(facecolor='pink', shrink=0.05))
plt.show()
```

```python
fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'],
 ['ax2', 'ax3']],
 figsize=(7, 4))
someaxes["ax1"].scatter(x1,y1)
someaxes["ax2"].bar(x2,y2)
someaxes["ax3"].plot(x1,y1)
someaxes["ax1"].set_xlabel('A Big Label',fontsize=18)
someaxes["ax2"].set_xlabel('Another Label',fontsize=18)
someaxes["ax3"].set_xlabel('Label 2: 2 Fast 2 Furious',fontsize=18 )
```

```python
fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'],
 ['ax2', 'ax3']],
 figsize=(20, 10), layout='constrained')
someaxes["ax1"].scatter(x1,y1)
someaxes["ax2"].bar(x2,y2)
someaxes["ax3"].plot(x1,y1)
someaxes["ax1"].set_xlabel('A Big Label',fontsize=18)
someaxes["ax2"].set_xlabel('Another Label',fontsize=18)
someaxes["ax3"].set_xlabel('Label 2: 2 Fast 2 Furious',fontsize=18 )
fig.tight_layout(pad=3.0)
```

```python
# first make our sample data
x = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y1 = np.array([110, 180, 240, 99, 220])
y2 = np.array([170, 100, 90, 120, 50])

# define our figure and axes (just one this time)
fig, ax = plt.subplots(figsize=(20, 10), constrained_layout=True)

# now call both bar and plot elements to the same axes (ax)
ax.bar(x, y1,
color = "blue")
ax.plot(x, y2,
color = "red")
label1 = 'Bar: Sales by Person'
label2 = 'Line: Customer Satisfaction'
ax.legend([label1, label2], loc='upper right')  
```

```python
y2_sd = np.std(y2)
```

```python
fig, ax = plt.subplots(figsize=(20, 10), constrained_layout=True)
ax.plot(x, y2, color = "red")
label2 = 'Line: Customer Satisfaction'
ax.legend([label2], loc='upper right')  
ax.fill_between(x, y2 - y2_sd, y2 + y2_sd, color='red', alpha=0.2)

ax.errorbar(x, #our x values
 y2, #our y values
 color = "red", #color of the central marker/line
 yerr = y2_sd,
 fmt = "none", #formatting for the central marker
 ecolor = "green", #color of the error bars
 elinewidth = 2, #thickness of the error bars
 capsize = 5, #length of the error bar caps
 label = "Customer Satisfaction with SD") #label for the legend
ax.legend([label2], loc='upper right')
```

```python
fig, ax = plt.subplots(figsize=(20, 10), constrained_layout=True)
ax.plot(x, y2,
 color = "red")
ax.errorbar(x,
 y2,
 yerr = y2_sd,
 fmt = "none",
 ecolor= "blue",
 elinewidth= 4,
 capsize = 6,
 capthick= 4 )
```
