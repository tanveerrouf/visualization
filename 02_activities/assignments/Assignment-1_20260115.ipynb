## Purpose

These font dictionaries are used with matplotlib's `fontdict` parameter to apply consistent and customizable text styling to plot titles, axis labels, and other text elements. This approach allows you to reuse the same font configurations across multiple plots without repeating code.

# Mirrored Density Plot Visualization

This code creates a mirrored density plot visualization with two probability distributions:

```python
# libraries
import numpy as np
from numpy import linspace
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import gaussian_kde

})

# Fig size
plt.rcParams["figure.figsize"]=12,8

sns.kdeplot(data=df, x="var1",  fill=True, alpha=1)

# plot density chart for var2
kde = gaussian_kde(df.var2)
x_range = linspace(min(df.var2), max(df.var2), len(df.var2))

# multiply by -1 to reverse axis (mirror plot)
sns.lineplot(x=x_range*-1, y=kde(x_range) * -1, color='orange') 
plt.fill_between(x_range*-1, kde(x_range) * -1, color='orange')

# add axis names        
plt.xlabel("value of x")
plt.axhline(y=0, linestyle='-',linewidth=1, color='black')

# show the graph
plt.show()
```

## Code Breakdown

- **Libraries**: Imports essential packages for numerical computing, data manipulation, and visualization.
- **DataFrame Creation**: Creates a dataframe with two variables:
  - **var1**: 1000 normally distributed values with default mean (0) and standard deviation (1).
  - **var2**: 1000 normally distributed values with mean of 2, multiplied by -1 for negative values.
- **Figure Size**: Sets the plot dimensions to 12 inches wide by 8 inches tall.
- **var1 Density Plot**: Uses seaborn's `kdeplot()` to create a filled kernel density estimate for var1.
- **var2 Density Plot**: Calculates kernel density estimate using `gaussian_kde()` and plots it with reversed axis (multiplied by -1).
- **Fill Area**: Adds an orange filled area under the var2 density curve.
- **Horizontal Line**: Adds a black horizontal line at y=0 to separate the two distributions.
- **Labels and Display**: Sets x-axis label and displays the final plot.

## Purpose

This visualization creates a mirror or back-to-back density plot, useful for comparing two probability distributions side by side. It helps identify similarities and differences between distributions and is commonly used in demographic or comparative analysis visualizations.

````python
# libraries
import numpy as np
from numpy import linspace
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import gaussian_kde

# dataframe
df = pd.DataFrame({
'var1': np.random.normal(size=1000),
'var2': np.random.normal(loc=2, size=1000) * -1
})

# Fig size
plt.rcParams["figure.figsize"]=12,8

# plot histogram chart for var1
sns.histplot(x=df.var1, stat="density", bins=20, edgecolor='black')

# plot histogram chart for var2
n_bins = 20
# get positions and heights of bars
heights, bins = np.histogram(df.var2, density=True, bins=n_bins) 
# multiply by -1 to reverse it
heights *= -1
bin_width = np.diff(bins)[0]
bin_pos =( bins[:-1] + bin_width / 2) * -1

# plot
plt.bar(bin_pos, heights, width=bin_width, edgecolor='black')

# show the graph
plt.show()
````
