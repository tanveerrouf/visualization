````python
np.random.seed(613)
x = np.arange(50)
y1 = np.random.randint(0, 100,50)
y2 = np.random.randint(0, 100,50)
````

# Generate Random Data with Multiple Variables

Sets a random seed for reproducibility and creates sample datasets with x-axis values (0-49) and two sets of random y-axis values (0-99).

## Code Breakdown

- **np.random.seed(613)**: Ensures reproducible random number generation.
- **x = np.arange(50)**: Creates an array of 50 integers from 0 to 49.
- **y1 = np.random.randint(0, 100, 50)**: Generates 50 random integers between 0 and 99.
- **y2 = np.random.randint(0, 100, 50)**: Generates another 50 random integers between 0 and 99.

## Purpose

Creates sample datasets useful for comparing two variables or plotting multiple series on the same chart.

````python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y1)
ax.plot(x,y2)
fig.show()
````

````python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y1, label = "Person 1" )
ax.plot(x,y2, label = "Person 2" )
ax.legend(loc='upper right')
fig.show()
````

# Legend Positioning

The legend position in matplotlib plots can be customized using the `loc` parameter in the `ax.legend()` function.

## Available Legend Positions

```python
ax.legend(loc='best')           # Automatically chooses the best position
ax.legend(loc='upper right')    # Top right corner
ax.legend(loc='upper left')     # Top left corner
ax.legend(loc='lower left')     # Bottom left corner
ax.legend(loc='lower right')    # Bottom right corner
ax.legend(loc='right')          # Right side, vertically centered
ax.legend(loc='center left')    # Left side, vertically centered
ax.legend(loc='center right')   # Right side, vertically centered
ax.legend(loc='lower center')   # Bottom center
ax.legend(loc='upper center')   # Top center
ax.legend(loc='center')         # Center of the plot
```

## Example Usage

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x, y1, label='Series 1')
ax.plot(x, y2, label='Series 2')
ax.legend(loc='upper right')    # Place legend in upper right corner
fig.show()
```

## Purpose

Customizing legend position helps avoid overlapping with data points and improves readability. Choose the position that best fits your data visualization without obscuring important information.

````python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y1, label = "Person 1")
ax.plot(x,y2, label = "Person 2")
ax.legend(loc='lower right',
 frameon = True, #add frame around the legend
 fontsize = 12, #change font size
 ncol = 2, #specify number of columns
 shadow = True )
fig.show()
````

````python
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1", color = 'red', marker = 'o' )
ax.scatter(x,y2, label = "Person 2", color = 'blue', marker = '^' )
ax.legend(loc='lower right')
ax.text(10, 95, "This value is important!")
fig.show()
````

````python
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1", color = 'red', marker = 'o' )
ax.scatter(x,y2, label = "Person 2", color = 'blue', marker = '^' )
ax.legend(loc='lower right')
ax.text(10, 95, "This value is important!", ha='center', fontsize=10, color='green', rotation=0,
        bbox=dict(facecolor='yellow', alpha=0.5))
````

````python
fig, ax = plt.subplots()
ax.axis([0, 10, 0, 10])

ax.text(1, 5, ". Data:(1, 5)", transform=ax.transData )
ax.text(0.5, 0.1, ". Axes:(0.5, 0.1)", transform=ax.transAxes )
ax.text(0.2, 0.2, ". Figure:(0.2, 0.2)", transform=fig.transFigure )
````

````python
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.annotate('This is important!', xy=(10, 95), xytext=(20, 94),
 arrowprops=dict(facecolor='black'))
fig.show()
````
